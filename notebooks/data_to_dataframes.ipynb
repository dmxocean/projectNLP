```python
import os
import sys
import pandas as pd
import pickle
import nltk

# Parent directory to PATH
sys.path.append("..")

from utils.json import load_json, save_json
from utils.dataset import AnnotationProcessor

from nltk.tokenize import TreebankWordTokenizer, sent_tokenize

from typing import Any
```

```python
# Define the paths
path_raw = "../data/raw"
path_annotations = "../data/annotations"
```

```python
# -- Process training data --
file_training = os.path.join(path_raw, "training.json")
data_training = load_json(file_training)
```

```python
# we use a class to ensure that the same tokenizer and stop words are used for all data
# it also allows us to change these parts easily
word_tokenizer = TreebankWordTokenizer()
sent_tokenizer_instance = nltk.data.load('tokenizers/punkt/english.pickle')
processor = AnnotationProcessor(word_tokenizer, sent_tokenizer_instance)
```

```python
# Convert training data to dataframe
df_train = processor.process_data(data_training)
print(f"Extracted {len(df_train)} annotations from training data")

# Save training dataframe
df_train_path = os.path.join(path_annotations, "df_train.pkl") 
with open(df_train_path, 'wb') as f:
    pickle.dump(df_train, f)
    
print(f"Saved training annotations to {df_train_path}")
```

```python
# -- Process test data --
file_test = os.path.join(path_raw, "test.json")
data_test = load_json(file_test)
print(f"Loaded test data with {len(data_test)} documents")
```

```python
# Convert test data to dataframe
df_test = processor.process_data(data_test, output_dir='../data/documents_test/')
print(f"Extracted {len(df_test)} annotations from test data")

# Save test dataframe
df_test_path = os.path.join(path_annotations, "df_test.pkl")
with open(df_test_path, 'wb') as f:
    pickle.dump(df_test, f)
print(f"Saved test annotations to {df_test_path}")
```

```python
print("Summary:")
print(f"Training documents: {len(data_training)}")
print(f"Training annotations: {len(df_train)}")
print()
print(f"Test documents: {len(data_test)}")
print(f"Test annotations: {len(df_test)}")
```
