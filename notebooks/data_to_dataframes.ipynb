```python
import os
import sys
import pandas as pd
import pickle
```

```python
# Parent directory to PATH
sys.path.append("..")

from utils.json import load_json, save_json
from utils.dataset import extract_annotations_and_split_documents
```

```python
path_data_raw = os.path.join("..", "data", "raw")
data_path_annotations = os.path.join("..", "data", "annotations")
```

```python
# -- Process training data --
file_training = os.path.join(path_data_raw, "training.json")
data_training = load_json(file_training)
```

```python
# Convert training data to dataframe
df_training_annotations = extract_annotations_and_split_documents(data_training, silent=True)
print(f"Extracted {len(df_training_annotations)} annotations from training data")

# Save training dataframe
df_training_annotations_path = os.path.join(data_path_annotations, "df_training_annotations_annotations.pkl")
with open(df_training_annotations_path, 'wb') as f:
    pickle.dump(df_training_annotations, f)
    
print(f"Saved training annotations to {df_training_annotations_path}")
```

```python
# -- Process test data --
file_test = os.path.join(path_data_raw, "test.json")
data_test = load_json(file_test)
print(f"Loaded test data with {len(data_test)} documents")
```

```python
# Convert test data to dataframe
df_test_annotations = extract_annotations_and_split_documents(data_test, silent=True)
print(f"Extracted {len(df_test_annotations)} annotations from test data")

# Save test dataframe
df_test_annotations_path = os.path.join(data_path_annotations, "df_test_annotations_annotations.pkl")
with open(df_test_annotations_path, 'wb') as f:
    pickle.dump(df_test_annotations, f)
print(f"Saved test annotations to {df_test_annotations_path}")
```

```python
print("Summary:")
print(f"Training documents: {len(data_training)}")
print(f"Training annotations: {len(df_training_annotations)}")
print()
print(f"Test documents: {len(data_test)}")
print(f"Test annotations: {len(df_test_annotations)}")
```
